```python
all_stores = pd.concat([
    Store_521,Store_632,Store_312,Store_323,Store_415

]
  ,axis=0                     )
all_stores = all_stores.reset_index(drop=True)

all_stores.head()
all_stores.shape
```

```python
all_stores['SKU'] = all_stores['SKU'].astype(str)
all_stores[all_stores['SKU'].str.contains('\\*')]
all_stores['SKU'] = all_stores['SKU'].str.replace('\\*' ,'0' ,regex=True)
```

```python
all_stores['Unit Sales'].value_counts(dropna=False)
```

```python
all_stores['Unit Sales'] = all_stores['Unit Sales'] .astype(str)

wrong_unit_sales = all_stores[all_stores['Unit Sales'].str.isnumeric()==False]
wrong_unit_sales
```

```python
all_stores['Unit Sales'].value_counts(dropna=False).tail(50)
```

```python
wrong_unit_sales['Unit Sales'].value_counts(dropna=False)
```

```python
all_stores['Unit Sales'] = all_stores['Unit Sales'].replace({
    '#0 0': '0',
    '#0 1': '1',
    '#0 2': '2',
    '#0 3': '3',
    '#0 4': '4',
    '#0 5': '5',

    '-A45%0 1': '1',
    '-A45%0 2': '2',
    '-A45%0 3': '3',
    '-A45%0 4': '4',
    '-A45%0 5': '5',
    '-A45%0 6': '6',
    '-A45%0 7': '7',
    '-A45%0 9': '9',

    '+9-A0': '90',
    '+9-A1': '91',
    '+9-A2': '92',
    '+9-A3': '93',

    '30000': '3',
    '40000': '4',
    '50000': '5'
})


all_stores['Unit Sales'].value_counts(dropna=False)
```

```python
all_stores['Unit Sales'] = all_stores['Unit Sales'].replace({'nan': np.nan, 'NaN' :np.nan})
all_stores['Unit Sales'] = all_stores['Unit Sales'].astype(float)
```
