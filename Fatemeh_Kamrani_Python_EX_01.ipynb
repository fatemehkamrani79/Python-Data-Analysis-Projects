Write a program to create a new string made of the middle three characters of an input string.

```python
# i am not sure whether this exercise is meant for strings with an even number of charachters or an odd number of characters.
```

```python
srt = input("Enter a string : ")
if  len(srt) % 2 ==0 :
    print("string length is even")
    mid=(len(srt))//2
    goal= srt[mid-2:mid+1] # If the string length is even, I choose the three middle characters from the left side.
    print(goal)
else :
    mid = len(srt)//2
    goal= srt[mid-1:mid+2]
    print(goal)
```
