```python
import pandas as pd
import numpy as np
```

```python
mydf = pd.read_excel ('TincyCo-V2.xlsx')
mydf
```

```python
mydf.columns
```

```python
mydf.info()
```

```python
mydf.isna().sum()
```

```python
dollar_mean = mydf['Dollar Sales'].mean()
dollar_mean
```

```python
mydf['Dollar Sales Mean Fill'] = mydf['Dollar Sales'].fillna(dollar_mean)
```

```python
mydf[['Dollar Sales','Dollar Sales Mean Fill']].isna().sum()
```

```python
mydf['Date'] = pd.to_datetime(mydf['Date'])
```

```python
mydf['Date'].dtype
```

```python
mydf['year'] = mydf['Date'].dt.year
mydf[['Date','year']].head()
```

```python
year_sales = mydf.groupby('year').agg(
    total_dollar_sales = ('Dollar Sales Mean Fill' , 'sum')
).reset_index()
year_sales
```

```python
mydf['quarter'] = mydf['Date'] .dt.quarter
mydf[['Date','year','quarter']].head()
```

```python
quarter_sales = mydf.groupby (
    ['year' ,'quarter']
).agg(
    total_dollar_sales = ('Dollar Sales Mean Fill' , 'sum')
    
).reset_index()
quarter_sales
```

```python
mydf['month'] = mydf ['Date'].dt.month
month_sales = mydf.groupby (
    ['year' ,'month']
).agg(
    total_dollar_sales = ('Dollar Sales Mean Fill','sum')
).reset_index()
month_sales
```

```python
mydf['week'] = mydf ['Date'].dt.isocalendar().week
```

```python
week_sales = mydf.groupby (
    ['year' ,'week']
).agg(
    total_dollar_sales = ('Dollar Sales Mean Fill','sum')
).reset_index()
week_sales
```

```python
mydf ['day_date'] = mydf['Date'].dt.date
```

```python
day_sales = mydf.groupby (
    'day_date'
).agg(
    total_dollar_sales = ('Dollar Sales Mean Fill','sum')
).reset_index()
day_sales 
```

```python
sku_month_sales = mydf.groupby (
    ['year' , 'month' ,'SKU']).agg(
        total_dollar_sales = ('Dollar Sales Mean Fill','sum')
    
    ).reset_index()
    
sku_month_sales
```

```python
sku_dollar_mean =mydf.groupby('SKU').agg(
    sku_mean_dollar_sales = ('Dollar Sales','mean')
).reset_index()
sku_dollar_mean.head()
```

```python
mydf = pd.merge(
    left= mydf,
    right= sku_dollar_mean,
    on='SKU',
    how='left'
)
mydf
```

```python
mydf[ 'Dollar Sales SKU Mean Fill'] = np.where(
    mydf['Dollar Sales'].isna(),
    mydf['sku_mean_dollar_sales'],
    mydf['Dollar Sales']
)

mydf
```
