```python
my_df = pd.read_excel('WoodInc-sale.xlsx')
my_df.rename(columns={'تاريخ':'date','کد مشتري' : 'customer_id','نام کالا':'productName','کد کالا':'productID','نام مشتري':'customerName','کد مشتري':'customerID','شماره':'docNumber','مبلغ خالص':'netAmount','ماليات':'tax','مبلغ':'amount','في':'unitPrice','مقدار':'quantity','واحد':'unit'}, inplace=True)
my_df.head()
```

```python
Coustomer_gp= mydf.groupby(['customerName','customerID']).agg(totalbuy=('netAmount','sum')).reset_index()
Coustomer_gp.head()
```

```python
Customer_gp = Customer_gp.sort_values(by='totalbuy',ascending=False).head(n=3)
Customer_gp.head(n=3)
```

```python
totalsales =mydf['netAmount'].sum()
totalsales
```

```python
topcustomer= Coustomer_gp.head(n=3)
topcustomer
```

```python
topcustomer ['share']=(topcustomer['totalbuy'] / totalsales)*100
topcustomer
```

```python
productgp= mydf.groupby(['productID' ,'productName']).agg(totalbuy=('netAmount','sum')).reset_index()
productgp=productgp.sort_values(by="totalbuy",ascending=False)

top_product =productgp.head(n=3)
top_product['share'] =(top_product['totalbuy']/totalsales)*100

top_product
```
