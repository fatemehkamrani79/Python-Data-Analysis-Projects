```python
my_df = pd.read_excel('WoodInc-sale.xlsx')
my_df.rename(columns={'تاريخ':'date','کد مشتري' : 'customer_id','نام کالا':'productName','کد کالا':'productID','نام مشتري':'customerName','کد مشتري':'customerID','شماره':'docNumber','مبلغ خالص':'netAmount','ماليات':'tax','مبلغ':'amount','في':'unitPrice','مقدار':'quantity','واحد':'unit'}, inplace=True)
my_df.head()
```

```python
my_df.info()
```

```python
my_df['customerID'] = my_df['customerID'].astype(str)
my_df['docNumber'] = my_df['docNumber'].astype(str)
my_df['date'] = my_df['date'].astype(str)
my_df.drop(columns=['customerID'], inplace=True)
my_df.info()
```

```python
my_df['customerName'].unique()
```

```python
my_df[my_df['customerName'].str.contains('ابوالفضل', na=False)]
my_df['customerName_clean'] = (
    my_df['customerName']
    .astype(str)
    .str.replace('ي', 'ی')
    .str.replace('ك', 'ک')
    .str.replace('\u200c', ' ')
    .str.replace('\u200f', '')
    .str.replace('\u200e', '')
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)
```
