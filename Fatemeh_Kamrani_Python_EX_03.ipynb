"Q1: Consider a triangle with vertices at (0, 0), (1, 0), and (0, 1). Write a function with the name my_inside_triangle(x,y) where the output is the string "outside" if the point (x, y) is outside of the triangle, the "border" if the point is exactly on the border of the triangle, and "inside" if the point is on the inside of the triangle."

```python
def my_inside_triangle(x, y):
    if x<0 or y<0 or x+y>1 :
        return "outside"
    elif x==0 or y==0 or x+y==1 :
        return "border"
    else:
        return "inside"
print(my_inside_triangle(5, 7))
print(my_inside_triangle(0.2, 0.3))
print(my_inside_triangle(0, 0))
```

"Q7: Turn the list words of lower case characters to upper case using the list comprehension"
 #words = ["test", "data", "analyze"]

```python
words = ["test", "data", "analyze"]
upper_words = [word.upper() for word in words]
print(upper_words)
```

"Q8: Find all of the numbers from 1–1000 that are divisible by 8 using list comprehension."
#[expression for item in iterable if condition]

```python
divisible_by_eight = [x for x in range(1,1001) if x%8 ==0]
print(divisible_by_eight)
```

"Q9: Find all of the numbers from 1–1000 that have a 6 in them using list comprehension."

```python
number_with_6 = [y for y in range(1,1001) if "6" in str(y) ]
print(number_with_6)
```

"Q11: Please search about dictionary comprehension and use it to count the length of each word in the following sentence."

```python
words = "Practice Problems to Drill Dictionary Comprehension in Your Head."
print(words.split())
words = words.replace(".", "")
result= {word: len(word) for word in words.split()}

print(result)
```
